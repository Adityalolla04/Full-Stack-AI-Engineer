```python
def run_instance_segmentation(model, image_tensor, score_threshold=0.5):
    """
    Run instance segmentation on an image
    Returns boxes, masks, labels, and scores
    """
    with torch.no_grad():
        predictions = model([image_tensor.to(device)])
    
    boxes = predictions[0]['boxes'].cpu().numpy()
    masks = predictions[0]['masks'].cpu().numpy()
    labels = predictions[0]['labels'].cpu().numpy()
    scores = predictions[0]['scores'].cpu().numpy()
    
    # Filter by score threshold
    mask = scores > score_threshold
    boxes = boxes[mask]
    masks = masks[mask]
    labels = labels[mask]
    scores = scores[mask]
    
    return boxes, masks, labels, scores

# Example usage with dummy image
dummy_image = torch.rand(3, 480, 640)
boxes, masks, labels, scores = run_instance_segmentation(model, dummy_image, score_threshold=0.5)

print(f"Detected instances: {len(boxes)}")
print(f"Masks shape: {masks.shape if len(masks) > 0 else 'Empty'}")
```
